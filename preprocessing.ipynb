```python
import pandas as pd

# membaca file csv
df = pd.read_csv("../data_cic/LDAP.csv", low_memory=False)
```

```python
# melihat 5 data teratas
df.head()
```

```python
df.shape
```

```python
X = df.drop(columns=[' Label'])
y = df[' Label']
```

```python
import numpy as np

X_numeric = X.select_dtypes(include=[np.number])

print("Jumlah fitur numerik:", X_numeric.shape[1])
```

```python
y_encoded = (
    y.replace({
        'BENIGN': 0,
        'LDAP': 1,
        'NetBIOS': 1
    })
    .infer_objects(copy=False)
    .astype(int)
)

print(y_encoded.value_counts())
```

```python
print(X_numeric.dtypes.unique())
print(y_encoded.dtype)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer

# ganti inf ke NaN dulu
X_numeric.replace([np.inf, -np.inf], np.nan, inplace=True)

imputer = SimpleImputer(strategy='median')
X_imputed = imputer.fit_transform(X_numeric)
```

```python
import pandas as pd

# hasil imputasi (array → DataFrame)
X_imputed_df = pd.DataFrame(
    X_imputed,
    columns=X_numeric.columns
)
```

```python
print(y_encoded.head())
```

```python
X_imputed_df.head()
```

```python
X_imputed_df = X_imputed_df.drop(columns=[' Source Port', ' Destination Port'])
```

```python
X_imputed_df.to_csv(
    "CICDDoS2019_LDAP_X.csv",
    index=False
)
```
